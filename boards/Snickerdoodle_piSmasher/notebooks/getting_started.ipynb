# snickerdoodle black + piSmasher: getting started

PYNQ 4.0 on a Zynq-7020. This notebook loads the base overlay, lists what it contains and allocates a contiguous buffer through XRT.

```python
from pynq import Overlay, allocate
import pynq

print('PYNQ', pynq.__version__)
ol = Overlay('base.bit')
print('Overlay loaded:', ol.is_loaded())
```

The base overlay is krtkl's piSmasher 4.1 HDMI pass-through design: both Ethernet ports bridged through the PL, HDMI in passed straight to HDMI out and the audio codec looped back. It has no AXI peripherals yet, so `ip_dict` is short.

```python
for name, ip in ol.ip_dict.items():
    print(name, ip['type'])
print()
for name, clk in ol.clock_dict.items():
    print(name, clk)
```

## Contiguous memory through XRT (zocl)

```python
import numpy as np

buf = allocate(shape=(1024,), dtype=np.uint32)
buf[:] = np.arange(1024, dtype=np.uint32)
buf.flush()
print('Physical address: 0x%08x' % buf.physical_address)
print('Sum:', int(buf.sum()))
buf.freebuffer()
```
